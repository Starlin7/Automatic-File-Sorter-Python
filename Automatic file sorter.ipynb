```python
import os,shutil
```

```python
path=r"C:/Users/Sharvin Pereira/OneDrive/Documents/sorter demo/"
```

```python
file_name=os.listdir(path)
```

```python
folder=["csv files","Image files","excel files"]
for i in range(0,3):
    if not os.path.exists(path + folder[i]):
        print(path + folder[i])
        os.makedirs(path + folder[i])
```

```python
for file in file_name:
    if ".csv" in file and not os.path.exists(path + "csv files/" + file):
        shutil.move(path + file,path + "csv files/" + file)
    if ".png" in file and not os.path.exists(path + "Image files/" + file):
        shutil.move(path + file,path + "Image files/" + file)
    if ".xlsx" in file and not os.path.exists(path + "excel files/" + file):
        shutil.move(path + file,path + "excel files/" + file)
        
```
